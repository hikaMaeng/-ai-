# 실습1 · 임베딩 진행 확인
`01_데이터적재_임베딩.ipynb` 가 임베딩을 채우는 동안 다른 탭에서 연다. 커널이 따로라 동시에 실행된다.
탭을 새로 고쳐 적재 노트북의 진행 출력이 멈춘 것처럼 보일 때도 여기서 실제 진행을 확인한다.

```python
# 셀 1 · 진행 확인 (다시 실행할 때마다 현재 값)
import psycopg

DB_URL = "postgresql://lab:lab@db:5432/lab"

with psycopg.connect(DB_URL) as conn:
    for t in ["docs", "questions"]:
        n, e = conn.execute(f"SELECT count(*), count(embedding) FROM {t}").fetchone()
        print(f"{t:9s} {e:6,} / {n:,}  ({e / n * 100:5.1f}%)" if n else f"{t:9s} 비어 있음")
```
